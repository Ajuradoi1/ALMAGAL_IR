```python
import pandas as pd
from astropy.table import Table
import numpy as np
```

```python
gut = Table.read('../doc/apjs300673t4_mrt.txt',format='ascii.cds')
```

```python
gut.rename_columns(['DE-','Jmag','e_Jmag','Hmag','e_Hmag','Kmag','e_Kmag','3.6mag','e_3.6mag','4.5mag','e_4.5mag','5.8mag','e_5.8mag','8.0mag','e_8.0mag','24mag','e_24mag','AK'],
                   ['DE','j_m','j_msigcom','h_m','h_msigcom','k_m','k_msigcom','mag3_6','d3_6m','mag4_5','d4_5m','mag5_8','d5_8m','mag8_0','d8_0m','mag24','d24m','Ak_NODEC'])


''' mag3_6| d3_6m|   mag4_5| d4_5m|   mag5_8| d5_8m|   mag8_0| d8_0m|       f_j|       f_h|      f_ks|      f3_6|     df3_6|      f4_5|    df4_5|      f5_8|     df5_8|      f8_0|     df8_0|color36_45|color36_58|color36_80|color45_58|color45_80|color58_80|has_alma|has_twomass|   j_m|j_msigcom|   h_m|h_msigcom|   k_m|k_msigcom|'''
```

```python
irac = ['36','45','58','80']
mag_irac = ['mag3_6','mag4_5','mag5_8','mag8_0']

for idx1,color in enumerate(irac):
    for idx2,color2 in enumerate(irac):
        if color != color2:
            gut['color'+color+'_'+color2] = gut[mag_irac[idx1]] - gut[mag_irac[idx2]]
```

```python
designation = np.zeros(len(gut))
for idx, element in enumerate(gut):
    designation[idx] = idx

gut['designation'] = designation
```

```python
gut.write('../doc/test_guthermuth09.txt',format='ascii.ecsv',overwrite=True)
```
