```python
import os

from astropy.table import Table
import numpy as np
```

```python
tbl = Table.read('../output/tables/match_table_glimpse_s07.tbl',format='ascii.ipac')
```

```python

# assuming your table is called tbl
col = tbl['region_id']

unique, counts = np.unique(col, return_counts=True)

# sort indices by counts (descending)
idx = np.argsort(counts)[::-1]

for u, c in zip(unique[idx], counts[idx]):
    print(u, c)
```
